# Colony: the language model (LFM3-1.7B + LoRA)

Fine-tunes Liquid AI's LFM3-1.7B with a LoRA adapter on Claude Haiku's answers to every language job in the game (goals, steps, talk, memories in people's own words, reading the player's words), and converts the adapter to GGUF for llama.cpp. Use the cheapest GPU: **Runtime > Change runtime type > T4 GPU**. Then **Runtime > Run all**. About an hour.

Nothing here needs a Hugging Face token (the model is public): if Colab asks for access to your secrets, say no.

The last cell zips what the game needs (`lm_game.zip`: `colony-lora.gguf` and `report.json`, tens of MB). Download it from the Files panel and put `colony-lora.gguf` in `brain/models/lm/` (renamed `colony-lora-v1.gguf`).
